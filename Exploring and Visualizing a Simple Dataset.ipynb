```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Nice plot style
sns.set(style="whitegrid")
%matplotlib inline
```

```python
# Change the path to where you saved your file
# Example: "datasets/iris.csv" or "C:/Users/YourName/Downloads/iris.csv"

df = pd.read_csv('iris.csv')   # ←←← CHANGE THIS TO YOUR ACTUAL FILE PATH
df.head()
```

```python
# Shape
print("Shape of the dataset:", df.shape)

# Column names
print("\nColumn names:")
print(df.columns.tolist())

# First 5 rows
print("\nFirst 5 rows:")
df.head()
```

```python
# Data types and missing values
df.info()

print("\n" + "="*50 + "\n")

# Numerical summary
df.describe()
```

```python
print("Missing values per column:")
print(df.isnull().sum())

print("\nSpecies distribution:")
print(df['species'].value_counts())   # change 'species' to your target column name if different
```

```python
# This one plot replaces many individual plots
sns.pairplot(df, hue='species', diag_kind='kde', corner=True, height=2.5)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='sepal-length', y='sepal-width', hue='species', style='species', s=100)
plt.title('Sepal Length vs Sepal Width')
plt.legend(title='Species')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='petal-length', y='petal-width', hue='species', style='species', s=100)
plt.title('Petal Length vs Petal Width')
plt.legend(title='Species')
plt.show()
```

```python
features = ['sepal-length', 'sepal-width', 'petal-length', 'petal-width']

plt.figure(figsize=(12, 8))
for i, feature in enumerate(features, 1):
    plt.subplot(2, 2, i)
    sns.histplot(data=df, x=feature, hue='species', kde=True, alpha=0.6)
    plt.title(f'Distribution of {feature}')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
for i, feature in enumerate(features, 1):
    plt.subplot(1, 4, i)
    sns.boxplot(data=df, y=feature, x='species')
    plt.title(feature)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
for i, feature in enumerate(features, 1):
    plt.subplot(1, 4, i)
    sns.violinplot(data=df, y=feature, x='species')
    plt.title(feature)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
# Select only numerical columns
numeric_df = df.select_dtypes(include=[np.number])
corr = numeric_df.corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', center=0, square=True)
plt.title('Feature Correlation Heatmap')
plt.show()
```
